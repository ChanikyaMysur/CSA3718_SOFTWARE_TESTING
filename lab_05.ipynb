```python
import unittest

# =====================================================================
# PART 1: FLIPKART PORTAL NEGATIVE SCENARIOS SIMULATION
# =====================================================================

class FlipkartNegativeExp5:
    """
    Simulates Flipkart behavior under negative test conditions as specified in Experiment 5.
    """
    def __init__(self):
        self.is_logged_in = True

    def login(self, email, password) -> str:
        """TC1: Incorrect email or password entered on Home login."""
        if email != "valid@email.com" or password != "correct_pass":
            return "Incorrect Email or password"
        return "Login Successful"

    def search_product(self, query) -> str:
        """TC2: Searching for products results in incorrect/wrong products appearing."""
        if query.lower() == "shirts":
            return "Wrong Products Appear (Shoes displayed instead)"
        return f"Results for {query}"

    def search_shops_mobiles(self) -> str:
        """TC3: Searching for mobiles under shops section fails to show them."""
        return "Searched Mobiles do not appear"

    def open_trending_stores_student_club(self) -> str:
        """TC4: Accessing student club under trending stores opens another store instead."""
        return "Student's Club does not open (Another store opened instead)"

    def check_notifications(self) -> str:
        """TC5: Order notification update does not show up in the notifications tab."""
        return "Does not show notifications"

    def access_help_centre(self) -> str:
        """TC6: Help Centre inside account details fails to appear."""
        return "Help centre Does not appear"

    def logout(self) -> str:
        """TC7: Clicking on logout option fails to log out the user."""
        return "Flipcart doesn't logout"

    def edit_profile(self) -> str:
        """TC8: Editing profile details fails to save or update."""
        return "Profile will not be updating"

    def place_order(self) -> str:
        """TC9: Order placement fails to execute."""
        return "Order not placed"

    def search_medicines(self) -> str:
        """TC10: Searching for medicines under shops category results in nothing found."""
        return "Medicines do not appear"


# =====================================================================
# PART 2: AUTOMATED NEGATIVE TEST SUITE & REPORT GENERATOR
# =====================================================================

class TestFlipkartNegativeExp5(unittest.TestCase):
    """
    Automated Negative Test Cases for Flipkart (Experiment 5).
    """

    def setUp(self):
        """Initialize system state before each negative test."""
        self.portal = FlipkartNegativeExp5()

    def print_test_report(self, item_no, item_name, features, steps, expected, actual):
        """Utility method to display structured laboratory test reports."""
        status = "Pass" if expected == actual else "Fail"
        print(f"\n{'='*70}")
        print(f"Item No         : {item_no}")
        print(f"Item to Test    : {item_name}")
        print(f"Features        : {features}")
        print(f"Test Steps      : {steps}")
        print(f"Expected Output : {expected}")
        print(f"Actual Output   : {actual}")
        print(f"Status          : {status}")
        print(f"{'='*70}")

    def test_TC_001_invalid_login(self):
        """Test Item 1: Home (Invalid Login)"""
        actual = self.portal.login("invalid@email.com", "wrong_password")
        expected = "Incorrect Email or password"
        self.print_test_report(
            item_no=1,
            item_name="Home (Search, Shops, Notifications, Account)",
            features="Email, Password validation",
            steps="1. Open Flipkart -> 2. Click on Email, password -> 3. Enter incorrect credentials",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_002_wrong_product_search(self):
        """Test Item 2: Search button wrong results"""
        actual = self.portal.search_product("shirts")
        expected = "Wrong Products Appear (Shoes displayed instead)"
        self.print_test_report(
            item_no=2,
            item_name="Search Button",
            features="Search for Brands & shirts",
            steps="1. Search for products -> 2. Observe incorrect product behavior",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_003_shops_mobiles_missing(self):
        """Test Item 3: Mobile Section under Shops"""
        actual = self.portal.search_shops_mobiles()
        expected = "Searched Mobiles do not appear"
        self.print_test_report(
            item_no=3,
            item_name="Shops (Mobiles, Fashion, Electronics)",
            features="Mobiles search inside Shops",
            steps="1. Open Shops -> 2. Click on Mobiles & search",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_004_student_club_trending_store(self):
        """Test Item 4: Student Club redirection issue"""
        actual = self.portal.open_trending_stores_student_club()
        expected = "Student's Club does not open (Another store opened instead)"
        self.print_test_report(
            item_no=4,
            item_name="Trending Stores",
            features="Search for student club and wedding store",
            steps="1. Open Shops -> 2. Navigate to Trending Stores -> 3. Click Student Club",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_005_notifications_failing(self):
        """Test Item 5: Notification delivery fail"""
        actual = self.portal.check_notifications()
        expected = "Does not show notifications"
        self.print_test_report(
            item_no=5,
            item_name="Notifications Panel",
            features="Order list notifications",
            steps="1. Open Notifications -> 2. Place an order -> 3. Verify notification existence",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_006_help_centre_missing(self):
        """Test Item 6: Account details Help Centre"""
        actual = self.portal.access_help_centre()
        expected = "Help centre Does not appear"
        self.print_test_report(
            item_no=6,
            item_name="Account (Orders, wishlist, coupons)",
            features="Help Centre navigation within Accounts",
            steps="1. Open Account -> 2. Select Help Centre",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_007_logout_failing(self):
        """Test Item 7: Account Logout Failure"""
        actual = self.portal.logout()
        expected = "Flipcart doesn't logout"
        self.print_test_report(
            item_no=7,
            item_name="Accounts (Logout Option)",
            features="Click on Logout",
            steps="1. Open Account -> 2. Scroll to bottom -> 3. Click logout",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_008_edit_profile_failing(self):
        """Test Item 8: Account Edit Profile Failure"""
        actual = self.portal.edit_profile()
        expected = "Profile will not be updating"
        self.print_test_report(
            item_no=8,
            item_name="Accounts (Edit Profile)",
            features="Click on edit profile",
            steps="1. Open Account -> 2. Click edit profile -> 3. Save updates",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_009_order_placement_failing(self):
        """Test Item 9: Cart Checkout Failure"""
        actual = self.portal.place_order()
        expected = "Order not placed"
        self.print_test_report(
            item_no=9,
            item_name="Cart Placement",
            features="Click on place order",
            steps="1. Open Accounts/Cart -> 2. Click Place Order",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_010_medicines_missing(self):
        """Test Item 10: Shops Category Medicines Missing"""
        actual = self.portal.search_medicines()
        expected = "Medicines do not appear"
        self.print_test_report(
            item_no=10,
            item_name="Shops (Medicines Category)",
            features="Search for medicines",
            steps="1. Open Shops -> 2. Select Medicines Category",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)


# =====================================================================
# PART 3: COLAB-COMPATIBLE TEST RUNNER
# =====================================================================
if __name__ == "__main__":
    print("\nExecuting Software Testing Lab - Experiment 5: Flipkart Portal Negative Test Suite\n")
    suite = unittest.TestLoader().loadTestsFromTestCase(TestFlipkartNegativeExp5)
    unittest.TextTestRunner(verbosity=1).run(suite)
```
