```python
import unittest

# =====================================================================
# PART 1: ARMS PORTAL SYSTEM SIMULATION (Target System under Test)
# =====================================================================

class ARMSPortal:
    """
    Simulates Saveetha ARMS Portal (arms.saveetha.com) core navigation,
    student dashboard, course management, and slot enrollment.
    """
    def __init__(self):
        # Database of authorized student accounts
        self.user_credentials = {
            "saveetha_student": "pass123"
        }
        self.logged_in_user = None

    def login(self, username, password) -> bool:
        """Simulates user authentication on arms.saveetha.com."""
        if username in self.user_credentials and self.user_credentials[username] == password:
            self.logged_in_user = username
            return True
        return False

    def get_home_dashboard(self) -> dict:
        """Retrieves Home tab data: Holidays Notification & Attendance Confirmation."""
        if not self.logged_in_user:
            return {"status": "FAIL", "message": "User not authenticated"}

        return {
            "status": "PASS",
            "holiday_notification": "Diwali Holiday Declared from Nov 12 - Nov 15",
            "attendance_confirmation": "Attendance Status: 94.5% Confirmed"
        }

    def get_my_courses(self) -> dict:
        """Retrieves My Course tab data: In progress, Graduation status, Completed courses, External credits."""
        if not self.logged_in_user:
            return {"status": "FAIL", "message": "User not authenticated"}

        return {
            "status": "PASS",
            "in_progress_courses": ["Software Testing Lab", "Cloud Computing"],
            "graduation_status": "Eligible for Graduation (Semester 7)",
            "completed_courses": ["Data Structures", "Operating Systems", "DBMS"],
            "external_credits": 6
        }

    def navigate_enrollment(self, selected_slot: str) -> dict:
        """Handles course enrollment by selecting Slot A, Slot B, Slot C, or Slot D."""
        if not self.logged_in_user:
            return {"status": "FAIL", "message": "User not authenticated"}

        valid_slots = ["Slot A", "Slot B", "Slot C", "Slot D"]
        if selected_slot not in valid_slots:
            return {"status": "FAIL", "message": "Invalid Slot Selected"}

        return {
            "status": "PASS",
            "selected_slot": selected_slot,
            "course_details": {
                "course_code": "CS1023",
                "course_name": "Software Quality Assurance and Testing",
                "credits": 4
            },
            "approval_note": "Approved"
        }


# =====================================================================
# PART 2: AUTOMATED TEST SUITE & CASE STUDY REPORT GENERATOR
# =====================================================================

class TestARMSPortal(unittest.TestCase):
    """
    Positive Test Automation Suite for ARMS Portal (Experiment 2).
    """

    def setUp(self):
        """Initialize system under test and log in before each test."""
        self.arms = ARMSPortal()
        self.username = "saveetha_student"
        self.password = "pass123"
        # Step 1 & 2 for all cases: Open site and enter credentials
        self.login_success = self.arms.login(self.username, self.password)

    def print_test_report(self, sl_no, item_name, features, approach, test_env, steps, status):
        """Prints formatted laboratory bug/test execution report."""
        print(f"\n{'='*70}")
        print(f"SL.No           : {sl_no}")
        print(f"Item to Test    : {item_name}")
        print(f"Features        : {features}")
        print(f"Approach        : {approach}")
        print(f"Test Environment: {test_env}")
        print(f"Test Steps      : {steps}")
        print(f"Status          : {status}")
        print(f"{'='*70}")

    def test_TC_001_home_holidays_attendance(self):
        """Test Item 1: Home Holidays Notification & Attendance Confirmation"""
        self.assertTrue(self.login_success, "Login failed before test execution.")

        home_res = self.arms.get_home_dashboard()

        is_holiday_valid = "holiday_notification" in home_res
        is_attendance_valid = "attendance_confirmation" in home_res

        actual_status = "Pass" if (home_res["status"] == "PASS" and is_holiday_valid and is_attendance_valid) else "Fail"

        self.print_test_report(
            sl_no=1,
            item_name="Home Holidays Notification & Attendance Confirmation",
            features="Holiday Notification, Attendance Confirmation",
            approach="Manual / Automated simulation",
            test_env="Operating system / Browser (arms.saveetha.com)",
            steps="1. Open website arms.saveetha.com -> 2. Enter username & password -> 3. Click home button -> 4. Check holiday notice and attendance confirmation.",
            status=actual_status
        )
        self.assertEqual(actual_status, "Pass")

    def test_TC_002_my_course(self):
        """Test Item 2: My Course Details Navigation"""
        self.assertTrue(self.login_success, "Login failed before test execution.")

        course_res = self.arms.get_my_courses()

        has_in_progress = "in_progress_courses" in course_res
        has_grad_status = "graduation_status" in course_res
        has_completed = "completed_courses" in course_res
        has_ext_credits = "external_credits" in course_res

        all_checks_passed = (course_res["status"] == "PASS" and has_in_progress and
                            has_grad_status and has_completed and has_ext_credits)
        actual_status = "Pass" if all_checks_passed else "Fail"

        self.print_test_report(
            sl_no=2,
            item_name="My Course",
            features="In progress course, Graduation status, Completed course, External credits",
            approach="Manual / Automated simulation",
            test_env="Operating system / Browser (arms.saveetha.com)",
            steps="1. Open website arms.saveetha.com -> 2. Enter username & password -> 3. Click '=' menu button -> 4. Select My Course.",
            status=actual_status
        )
        self.assertEqual(actual_status, "Pass")

    def test_TC_003_enrollment(self):
        """Test Item 3: Course Enrollment Slot A Selection"""
        self.assertTrue(self.login_success, "Login failed before test execution.")

        enroll_res = self.arms.navigate_enrollment("Slot A")

        slot_matched = enroll_res.get("selected_slot") == "Slot A"
        has_course_details = "course_details" in enroll_res
        has_approval = enroll_res.get("approval_note") in ["Approved", "Pending"]

        actual_status = "Pass" if (enroll_res["status"] == "PASS" and slot_matched and has_course_details and has_approval) else "Fail"

        self.print_test_report(
            sl_no=3,
            item_name="Enrollment",
            features="Select Slot A, Slot B, Slot C, Slot D, Course details, Note (Approved/Pending)",
            approach="Manual / Automated simulation",
            test_env="Operating system / Browser (arms.saveetha.com)",
            steps="1. Open website arms.saveetha.com -> 2. Enter username & password -> 3. Click '=' menu button -> 4. Go to Enrollment -> 5. Select Slot A.",
            status=actual_status
        )
        self.assertEqual(actual_status, "Pass")


# =====================================================================
# PART 3: COLAB-COMPATIBLE TEST RUNNER
# =====================================================================
if __name__ == "__main__":
    print("\nExecuting Software Testing Lab - Experiment 2: ARMS Portal Positive Test Cases\n")
    unittest.main(argv=[''], exit=False)
```
