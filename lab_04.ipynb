```python
import unittest

# =====================================================================
# PART 1: E-COMMERCE PORTAL (FLIPKART) NEGATIVE SCENARIOS SIMULATION
# =====================================================================

class FlipkartNegative:
    """
    Simulates e-Commerce (Flipkart) behavior under negative/unauthenticated test conditions,
    handling invalid interactions across various features.
    """
    def __init__(self):
        self.is_logged_in = False
        self.cart = []
        self.addresses = []
        self.orders = []

    def login(self, username, password) -> str:
        """TC1: Handles invalid login authentication."""
        valid_user = "customer_user"
        valid_pass = "secure_pass"
        if username == valid_user and password == valid_pass:
            self.is_logged_in = True
            return "Login Successful"
        return "Invalid username or password"

    def search_product(self, query) -> str:
        """TC2: Negative search handling for empty/invalid query."""
        if not query or len(query.strip()) == 0:
            return "Please enter a valid product name"
        if query.lower() in ["illegal", "restricted_item"]:
            return "No matching products found"
        return f"Showing results for {query}"

    def view_shops_offers(self) -> str:
        """TC3: Accessing exclusive member shops without authentication."""
        if not self.is_logged_in:
            return "Access Denied. Please login to view personalized offers"
        return "Displaying Trending Stores and Offers"

    def get_notifications(self) -> str:
        """TC4: Attempting to pull system notifications when logged out."""
        if not self.is_logged_in:
            return "No notifications available. User not logged in."
        return "Displaying recent notification alerts"

    def get_account_details(self) -> str:
        """TC5: Fetching user orders or wishlist with a failed session."""
        if not self.is_logged_in:
            return "Session expired. Cannot fetch account information"
        return "User Account Details Loaded"

    def add_to_cart(self, product) -> str:
        """TC6: Attempting to add an invalid/out of stock product to the cart."""
        out_of_stock_products = ["iPhone 12", "PlayStation 5"]
        if product in out_of_stock_products:
            return "Product is out of stock"
        self.cart.append(product)
        return "Product added to cart successfully"

    def edit_profile(self, phone) -> str:
        """TC7: Attempting to update profile with invalid format input."""
        if not self.is_logged_in:
            return "Authentication required to update profile"
        if not phone.isdigit() or len(phone) != 10:
            return "Failed to update. Invalid phone number format"
        return "Profile Updated Successfully"

    def remove_from_cart(self, product) -> str:
        """TC8: Attempting to remove a product that does not exist in cart."""
        if product not in self.cart:
            return "Item not found in your cart"
        self.cart.remove(product)
        return "Item removed successfully"

    def place_order(self) -> str:
        """TC9: Attempting to place order with an empty cart or missing address."""
        if not self.is_logged_in:
            return "Login required to place order"
        if not self.cart:
            return "Cart is empty. Add products to place an order"
        if not self.addresses:
            return "Delivery address is required"
        return "Order Placed Successfully"

    def cancel_order(self, order_id) -> str:
        """TC10: Attempting to cancel an order that does not exist or has already shipped."""
        shipped_orders = ["ORD101", "ORD102"]
        if order_id in shipped_orders:
            return "Cannot cancel order. Already shipped"
        if order_id not in self.orders:
            return "Order ID not found"
        self.orders.remove(order_id)
        return "Order Cancelled Successfully"


# =====================================================================
# PART 2: AUTOMATED NEGATIVE TEST SUITE FOR E-COMMERCE PORTAL
# =====================================================================

class TestFlipkartNegative(unittest.TestCase):
    """
    Automated Negative Test Cases for Flipkart (Experiment 4).
    """

    def setUp(self):
        """Initialize system state before each negative test."""
        self.portal = FlipkartNegative()

    def print_test_report(self, item_no, item_name, features, steps, expected, actual):
        """Utility method to display structured laboratory test reports."""
        status = "Pass" if expected == actual else "Fail"
        print(f"\n{'='*70}")
        print(f"Item No         : {item_no}")
        print(f"Item to Test    : {item_name}")
        print(f"Features        : {features}")
        print(f"Test Steps      : {steps}")
        print(f"Expected Output : {expected}")
        print(f"Actual Output   : {actual}")
        print(f"Status          : {status}")
        print(f"{'='*70}")

    def test_TC_001_invalid_login(self):
        """Test Item 1: Home (Invalid Login)"""
        actual = self.portal.login("customer_user", "wrong_pass")
        expected = "Invalid username or password"
        self.print_test_report(
            item_no=1,
            item_name="Home Login Verification",
            features="Email/Username, Password Validation",
            steps="1. Open Portal -> 2. Enter correct username -> 3. Enter incorrect password -> 4. Click login",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_002_search_empty_query(self):
        """Test Item 2: Search Button (Empty Query)"""
        actual = self.portal.search_product("   ")
        expected = "Please enter a valid product name"
        self.print_test_report(
            item_no=2,
            item_name="Search Functionality",
            features="Empty Input Handling",
            steps="1. Select Search field -> 2. Input empty space characters -> 3. Click search",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_003_shops_unauthorized(self):
        """Test Item 3: Shops and Deals (Unauthorized Access)"""
        actual = self.portal.view_shops_offers()
        expected = "Access Denied. Please login to view personalized offers"
        self.print_test_report(
            item_no=3,
            item_name="Shops & Trending Stores",
            features="Authentication Check for Personalized Offers",
            steps="1. Open homepage without log in -> 2. Click Shops/Offers Section -> 3. Read status response",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_004_notifications_unauthorized(self):
        """Test Item 4: Notifications (Unauthorized Access)"""
        actual = self.portal.get_notifications()
        expected = "No notifications available. User not logged in."
        self.print_test_report(
            item_no=4,
            item_name="Notifications Panel",
            features="Private Alerts and Promo Distribution Control",
            steps="1. Open homepage -> 2. Click on Notifications while signed out -> 3. Verify restriction message",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_005_account_details_unauthorized(self):
        """Test Item 5: Account Section (Session Expired)"""
        actual = self.portal.get_account_details()
        expected = "Session expired. Cannot fetch account information"
        self.print_test_report(
            item_no=5,
            item_name="Account Page Dashboard",
            features="Wishlist, Orders and Edit Profile Restriction",
            steps="1. Access profile settings without login session -> 2. Try fetching personal data",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_006_add_to_cart_out_of_stock(self):
        """Test Item 6: Cart (Out of Stock Product)"""
        actual = self.portal.add_to_cart("iPhone 12")
        expected = "Product is out of stock"
        self.print_test_report(
            item_no=6,
            item_name="Add to Cart Handler",
            features="Inventory Availability Validation",
            steps="1. Navigate to out-of-stock product page -> 2. Click on 'Add to Cart'",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_007_edit_profile_invalid_phone(self):
        """Test Item 7: Account Settings (Invalid Phone Number)"""
        self.portal.login("customer_user", "secure_pass")
        actual = self.portal.edit_profile("98765abcde")
        expected = "Failed to update. Invalid phone number format"
        self.print_test_report(
            item_no=7,
            item_name="Account Settings Profile Edit",
            features="Profile Form Validation",
            steps="1. Login -> 2. Access edit profile -> 3. Enter alphanumeric/invalid phone number -> 4. Submit",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_008_remove_unlisted_product_from_cart(self):
        """Test Item 8: My Cart (Removing Unlisted Product)"""
        actual = self.portal.remove_from_cart("Laptop Adapter")
        expected = "Item not found in your cart"
        self.print_test_report(
            item_no=8,
            item_name="My Cart Modifications",
            features="Dynamic Item Removal Validation",
            steps="1. View cart page -> 2. Attempt to trigger removal event for an item not added",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_009_place_order_empty_cart(self):
        """Test Item 9: Order Placement (Empty Cart)"""
        self.portal.login("customer_user", "secure_pass")
        actual = self.portal.place_order()
        expected = "Cart is empty. Add products to place an order"
        self.print_test_report(
            item_no=9,
            item_name="Order Placement Gateway",
            features="Checkout Preconditions Validation",
            steps="1. Login -> 2. Leave cart empty -> 3. Proceed directly to checkout and click place order",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_010_cancel_shipped_order(self):
        """Test Item 10: Account (Cancel Shipped Order)"""
        self.portal.login("customer_user", "secure_pass")
        actual = self.portal.cancel_order("ORD101")
        expected = "Cannot cancel order. Already shipped"
        self.print_test_report(
            item_no=10,
            item_name="Order Lifecycle Actions",
            features="Cancel Order Constraints",
            steps="1. View user order history -> 2. Attempt cancellation of an order marked 'shipped'",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)


# =====================================================================
# PART 3: COLAB-COMPATIBLE TEST RUNNER
# =====================================================================
if __name__ == "__main__":
    print("\nExecuting Software Testing Lab - Experiment 4: Flipkart Portal Negative Test Suite\n")
    # Run suite
    suite = unittest.TestLoader().loadTestsFromTestCase(TestFlipkartNegative)
    unittest.TextTestRunner(verbosity=1).run(suite)
```
