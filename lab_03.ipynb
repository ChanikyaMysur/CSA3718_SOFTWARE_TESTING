```python
import unittest

# =====================================================================
# PART 1: ARMS PORTAL NEGATIVE SCENARIOS SIMULATION
# =====================================================================

class ARMSPortalNegative:
    """
    Simulates ARMS Portal behavior under negative test conditions,
    unauthenticated access, invalid credentials, and system restriction states.
    """
    def __init__(self):
        self.is_logged_in = False

    def login(self, username, password) -> str:
        """TC1: Handles invalid login authentication."""
        valid_user = "student_user"
        valid_pass = "correct_password"

        if username == valid_user and password == valid_pass:
            self.is_logged_in = True
            return "Login Successful"
        return "Username and password entered is invalid"

    def get_my_courses(self) -> str:
        """TC2: Attempts to fetch course details under failed session."""
        if not self.is_logged_in:
            return "My course wasn't appeared"
        return "Courses Displayed Successfully"

    def get_enrollment(self) -> str:
        """TC3: Attempts to access slot enrollment under failed session."""
        if not self.is_logged_in:
            return "My enrollment wasn't appeared"
        return "Enrollment Slots Available"

    def get_attendance_report(self) -> str:
        """TC4: Attempts to fetch attendance report without valid session."""
        if not self.is_logged_in:
            return "Attendance wasn't appeared"
        return "Attendance Report Loaded"

    def open_assignment(self) -> str:
        """TC5: Attempts to open course assignment under unverified access."""
        if not self.is_logged_in:
            return "Assignment wasn't appeared"
        return "Assignment Opened"

    def process_no_due_request(self) -> str:
        """TC6: Examination No Due Request processing failure."""
        if not self.is_logged_in:
            return "Request is not accepted"
        return "No Due Request Approved"

    def check_financial_record(self) -> str:
        """TC7: Financial record due list and payment history check."""
        if not self.is_logged_in:
            return "Payment history not cleared"
        return "Payment History Cleared"

    def get_disciplinary_issue_details(self) -> str:
        """TC8: Disciplinary issue resolution check."""
        if not self.is_logged_in:
            return "Issues were not cleared"
        return "All Disciplinary Issues Cleared"

    def upload_infra_issue_file(self) -> str:
        """TC9: Infrastructure issue upload file path check."""
        if not self.is_logged_in:
            return "Not showing file location"
        return "File Location Selected Successfully"

    def update_student_profile(self) -> str:
        """TC10: Edit student profile and record update check."""
        if not self.is_logged_in:
            return "Profile was not updating"
        return "Profile Updated Successfully"


# =====================================================================
# PART 2: AUTOMATED NEGATIVE TEST SUITE & REPORT GENERATOR
# =====================================================================

class TestARMSPortalNegative(unittest.TestCase):
    """
    Automated Negative Test Cases for ARMS Portal (Experiment 3).
    """

    def setUp(self):
        """Initialize system instance before each test (unauthenticated state)."""
        self.arms = ARMSPortalNegative()

    def print_test_report(self, item_no, item_name, features, steps, expected, actual):
        """Utility method to display structured laboratory test reports."""
        status = "Pass" if expected == actual else "Fail"
        print(f"\n{'='*70}")
        print(f"Item No         : {item_no}")
        print(f"Item to Test    : {item_name}")
        print(f"Features        : {features}")
        print(f"Test Steps      : {steps}")
        print(f"Expected Output : {expected}")
        print(f"Actual Output   : {actual}")
        print(f"Status          : {status}")
        print(f"{'='*70}")

    def test_TC_001_invalid_login(self):
        """Test Item 1: Invalid Login Authentication"""
        actual = self.arms.login("student_user", "wrong_pass")
        expected = "Username and password entered is invalid"
        self.print_test_report(
            item_no=1,
            item_name="Home (Holidays, Notification, Attendance)",
            features="Username, Password",
            steps="1. Open website -> 2. Enter username & password -> 3. Enter wrong password",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_002_my_course(self):
        """Test Item 2: My Course Access Failure"""
        actual = self.arms.get_my_courses()
        expected = "My course wasn't appeared"
        self.print_test_report(
            item_no=2,
            item_name="My Course",
            features="In Progress Courses, External credit status",
            steps="1. Click '=' button -> 2. Check whether completed courses is there or not",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_003_enrollment(self):
        """Test Item 3: Enrollment Slot Access Failure"""
        actual = self.arms.get_enrollment()
        expected = "My enrollment wasn't appeared"
        self.print_test_report(
            item_no=3,
            item_name="Enrollment",
            features="Select Slot A, Slot B, Slot C, Course Details, Note: Approved",
            steps="1. Click '=' button -> 2. Go to enrollment -> 3. Select Slot A, B, C",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_004_attendance(self):
        """Test Item 4: Attendance Report View Failure"""
        actual = self.arms.get_attendance_report()
        expected = "Attendance wasn't appeared"
        self.print_test_report(
            item_no=4,
            item_name="Attendance",
            features="S. No, Course Code, Course Name, Class attended view",
            steps="1. Click '=' button -> 2. Go to Attendance -> 3. Click on Attendance report",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_005_assignment(self):
        """Test Item 5: Opening Assignment Failure"""
        actual = self.arms.open_assignment()
        expected = "Assignment wasn't appeared"
        self.print_test_report(
            item_no=5,
            item_name="Assignment",
            features="Completed Assignment, content, Exams",
            steps="1. Click '≡' button -> 2. Go on Assignment -> 3. Select Course opening Assignment",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_006_examination(self):
        """Test Item 6: Examination No Due Request Failure"""
        actual = self.arms.process_no_due_request()
        expected = "Request is not accepted"
        self.print_test_report(
            item_no=6,
            item_name="Examination",
            features="No due request, popup options for hall ticket",
            steps="1. Click '≡' button -> 2. Click on examination -> 3. Go to no due request",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_007_financial_record(self):
        """Test Item 7: Financial Record Due List Failure"""
        actual = self.arms.check_financial_record()
        expected = "Payment history not cleared"
        self.print_test_report(
            item_no=7,
            item_name="Financial Record",
            features="Due List, Payment history, Print Payment",
            steps="1. Click '≡' button -> 2. Go to Financial Record -> 3. Check due list, Payment history",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_008_disciplinary(self):
        """Test Item 8: Disciplinary Record Issue Failure"""
        actual = self.arms.get_disciplinary_issue_details()
        expected = "Issues were not cleared"
        self.print_test_report(
            item_no=8,
            item_name="Disciplinary",
            features="Disciplinary Record, Issue details, Last action details",
            steps="1. Click '≡' button -> 2. Go to Disciplinary issue details -> 3. Last action details",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_009_raise_infra_issue(self):
        """Test Item 9: Infrastructure Issue Location File Failure"""
        actual = self.arms.upload_infra_issue_file()
        expected = "Not showing file location"
        self.print_test_report(
            item_no=9,
            item_name="Raise Infra Issue",
            features="Raise issue content, choose file location",
            steps="1. Click '≡' button -> 2. Go to infra issues content -> 3. Choose file location",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)

    def test_TC_010_my_profile(self):
        """Test Item 10: Student Profile Update Failure"""
        actual = self.arms.update_student_profile()
        expected = "Profile was not updating"
        self.print_test_report(
            item_no=10,
            item_name="My Profile",
            features="Edit profile, student record details",
            steps="1. Click '≡' button -> 2. Go on my profile -> 3. Check student record details",
            expected=expected,
            actual=actual
        )
        self.assertEqual(expected, actual)


# =====================================================================
# PART 3: COLAB-COMPATIBLE TEST RUNNER
# =====================================================================
if __name__ == "__main__":
    print("\nExecuting Software Testing Lab - Experiment 3: ARMS Portal Negative Test Suite\n")
    # argv=[''] prevents Colab CLI flags from breaking unittest
    # exit=False keeps Colab kernel alive after test execution
    unittest.main(argv=[''], exit=False)
```
